```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
data = pd.read_csv("Downloads/1730285881-Airbnb_Open_Data.csv", low_memory= False)
data.head()
```

```python
data.info()
```

```python
data.duplicated().value_counts()
```

```python
data[data["neighbourhood group"] == "brookln"]
```

```python
data.drop_duplicates(inplace=True)
```

```python
data.drop(["house_rules","license"], axis = 1, inplace =True)
```

```python
data['price'] = data['price'].str.replace("$","",regex=False)
data['service fee'] = data['service fee'].str.replace("$","",regex=False)
```

```python
data['price'] = data['price'].str.replace(",","",regex=False)
data['service fee'] = data['service fee'].str.replace(",","",regex=False)
```

```python
data.rename(columns = {
    "price" : "price_$",
    "service fee" : "service_fee_$"
}, inplace = True)
```

```python
data.dropna(inplace = True)
```

```python
data['price_$'] = data['price_$'].astype(float)
data['service_fee_$'] = data['service_fee_$'].astype(float)
data['id'] = data['id'].astype(str)
data['host id'] = data['host id'].astype(str)
data['last review'] = pd.to_datetime(data['last review'])
data['Construction year'] = data['Construction year'].astype(int)
```

```python
data.loc[data['neighbourhood group'] == 'brookln','neighbourhood group'] = 'Brooklyn'
```

```python
data = data.drop(data[data["availability 365"]> 500].index)
```

```python
data.duplicated().value_counts()
```

```python
data.info()
```

```python
data.describe()
```

```python
property_type = data["room type"].value_counts().to_frame()
property_type
```

```python
room_type_bar = plt.bar(property_type.index, property_type.loc[:,"count"]);
plt.bar_label(room_type_bar, label = property_type.loc[:,"count"] , padding =4);
plt.ylim(0,50000);
plt.xlabel("Room type")
plt.ylabel("Room type count")
plt.title("Property type and their count in the dataset")
```

```python
hood_group = data["neighbourhood group"].value_counts().to_frame()
hood_group
```

```python
hood_group_bar = plt.bar(hood_group.index, hood_group.loc[:,"count"]);
plt.bar_label(hood_group_bar, label = hood_group.loc[:,"count"] , padding =4);
plt.ylim(0,40000);
plt.xlabel("Neighbourhood Group");
plt.ylabel("Number of listing");
plt.xticks(rotation = 45);
plt.title("Which Neighbourhood group is highest number of listing");
```

```python
avg_price = data.groupby("neighbourhood group")['price_$'].mean().sort_values(ascending = False).to_frame()
avg_price_bar = plt.bar(avg_price.index, avg_price.loc[:,"price_$"]);
plt.bar_label(avg_price_bar, label = round(avg_price.loc[:,"price_$"]), label_type = "edge" , padding =4);
plt.ylim(0,700);
plt.xlabel("Neighbourhood Group");
plt.ylabel("Average price per listing ($)");
plt.xticks(rotation = 45);
plt.title("Average price per listing ($) in each Neighbourhood Group");
```

```python
data.groupby(data["Construction year"])['price_$'].mean().to_frame().plot();
plt.xlabel("Construction year");
plt.ylabel("Average price($)");
plt.title("Average price($) for properties in each Construction year");
```

```python
hosts= data.groupby("host name")['calculated host listings count'].sum().sort_values(ascending = False).nlargest(10).to_frame()
hosts_bar = plt.bar(hosts.index, hosts.loc[:,"calculated host listings count"]);
plt.bar_label(hosts_bar, label = hosts.loc[:,"calculated host listings count"], label_type = "edge" , padding = 3);
plt.xlabel("host name");
plt.ylabel("calculate host listings count");
plt.xticks(rotation = 80);
plt.ylim([0,120000]);
plt.title("Top 10 hosts by calculate host listings count ");
```

```python
review = data.groupby("host_identity_verified")["review rate number"].mean().sort_values(ascending = False).to_frame()
review
```

```python
review_bar = plt.bar(review.index, review.loc[:,"review rate number"]);
plt.bar_label(review_bar, label = round(review.loc[:,"review rate number"],2), label_type = "edge" , padding =4);
plt.ylim(0,4);
plt.xlabel("Host Verification Status");
plt.ylabel("Average review rate number($)");
plt.title("Average Review Rate for each Verification Status");
```

```python
base_color = sns.color_palette()[0]
sns.boxplot(data = data, x ="host_identity_verified", y = "review rate number",color = base_color);
plt.xlabel("Host Verification Status");
plt.ylabel("Average review rate number($)");
plt.title("Average Review Rate for each Verification Status");
```

```python
data["price_$"].corr(data["service_fee_$"])
```
